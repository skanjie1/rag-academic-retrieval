# Corpus Exploration
Exploratory analysis of 537 NLP paper abstracts.

```python
import json, pandas as pd, matplotlib.pyplot as plt
with open("../data/nlp_abstracts.json") as f: papers = json.load(f)
df = pd.DataFrame(papers)
print(f"Corpus: {len(df)} papers")
df["year"].value_counts().sort_index().plot(kind="bar", title="Papers by Year")
plt.show()
```
