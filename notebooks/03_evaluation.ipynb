# RAGAS Evaluation Analysis
Compare baseline vs fine-tuned retrieval quality.

```python
import json, pandas as pd, matplotlib.pyplot as plt
with open("../evaluation/results.json") as f: results = json.load(f)
comp = pd.DataFrame({"Metric": ["Ctx Precision","Faithfulness","Answer Rel.","Ctx Recall"], "Baseline": list(results["baseline"].values()), "Fine-tuned": list(results["finetuned"].values())})
comp["Delta"] = comp["Fine-tuned"] - comp["Baseline"]
print(comp.to_string(index=False))
```
